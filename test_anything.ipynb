```python
from dowhy import CausalModel
import dowhy.datasets
 
# Load example data
data = dowhy.datasets.linear_dataset(
    beta=10,
    num_common_causes=5,
    num_instruments=2,
    num_samples=10000,
    treatment_is_binary=True)
 
# 1. Modeling: create the causal model
model = CausalModel(
    data=data["df"],
    treatment=data["treatment_name"],
    outcome=data["outcome_name"],
    graph=data["gml_graph"])
 
# 2. Identification: determine whether the causal effect is identifiable
identified_estimand = model.identify_effect()
 
# 3. Estimation: estimate the causal effect using statistical methods
estimate = model.estimate_effect(identified_estimand,
                                 method_name="backdoor.propensity_score_matching")
 
# 4. Refutation: run multiple robustness checks on the estimate
refute_results = model.refute_estimate(identified_estimand, estimate,
                                       method_name="random_common_cause")

print("Sample rows (treatment, outcome, causes, instruments):")
display(data["df"].head())
print("\nIdentified estimand:")
print(identified_estimand)
print("\nEstimated average treatment effect (ATE):")
print(estimate)
print("\nRefutation via random common cause:")
print(refute_results)
```

```python
# Try graphviz visualization first
graphviz_success = False
try:
    from IPython.display import Image
    import os
    graph_path = model.view_model(layout="dot")
    if graph_path and os.path.exists(graph_path):
        display(Image(filename=graph_path))
        print("Causal graph displayed successfully using Graphviz!")
        graphviz_success = True
    else:
        print("Graphviz path not found")
except Exception as e:
    print(f"Graphviz visualization failed: {e}")

# If graphviz failed, try alternative visualization
if not graphviz_success:
    print("\nTrying alternative visualization with NetworkX...")
    try:
        import networkx as nx
        import matplotlib.pyplot as plt
        from networkx.drawing.nx_agraph import read_dot

        # Try to use the GML graph from the data
        if "gml_graph" in data:
            # Parse GML and create network
            G = nx.parse_gml(data["gml_graph"])

            # Create visualization
            plt.figure(figsize=(12, 8))
            pos = nx.spring_layout(G, seed=42)

            # Color nodes based on their role
            node_colors = []
            node_labels = {}
            for node in G.nodes():
                if isinstance(data["treatment_name"], list) and str(node) in [str(t) for t in data["treatment_name"]]:
                    node_colors.append('lightblue')  # treatment
                    node_labels[node] = f"{node}\n(Treatment)"
                elif str(node) == str(data["outcome_name"]):
                    node_colors.append('lightgreen')  # outcome
                    node_labels[node] = f"{node}\n(Outcome)"
                else:
                    node_colors.append('lightgray')  # confounders/instruments
                    node_labels[node] = f"{node}\n(Variable)"

            nx.draw(G, pos, labels=node_labels, node_color=node_colors,
                   node_size=3000, font_size=9, font_weight='bold',
                   arrows=True, arrowsize=20, edge_color='gray',
                   bbox=dict(boxstyle="round,pad=0.3", facecolor="white", alpha=0.8))

            plt.title("Causal Graph Visualization\nBlue: Treatment, Green: Outcome, Gray: Other Variables", fontsize=14)
            plt.axis('off')
            plt.tight_layout()
            plt.show()

            print("Alternative causal graph visualization completed!")
        else:
            print("GML graph not available in data")

    except Exception as alt_e:
        print(f"Alternative visualization failed: {alt_e}")
        print("\nManual graph inspection:")
        try:
            print("Available data keys:", list(data.keys()))
            if "gml_graph" in data:
                print("GML graph (first 200 chars):", data["gml_graph"][:200])
        except Exception as inspect_e:
            print(f"Could not inspect graph: {inspect_e}")
```
